# Cybersecurity and Cryptography

**Cybersecurity** is the practice of protecting systems, networks, and programs from digital attacks. These attacks are usually aimed at accessing, changing, or destroying sensitive information; extorting money from users; or interrupting normal business processes.

**Cryptography** is a fundamental component of cybersecurity. It is the practice and study of techniques for secure communication in the presence of adversarial behavior. Essentially, it's about transforming information into a secure format, making it unreadable to unauthorized individuals. This includes encryption (converting plaintext into ciphertext) and decryption (converting ciphertext back into plaintext), as well as digital signatures and secure hashing, ensuring data confidentiality, integrity, and authenticity.



## Symmetric Cryptography

**Symmetric cryptography**, also known as secret-key cryptography, is a method of encryption where the same secret key is used for both encrypting and decrypting the information. This means that both the sender and the receiver must possess this identical key to communicate securely. Symmetric algorithms are generally much faster than asymmetric algorithms and are commonly used for bulk encryption of data. However, the secure exchange of the secret key itself is a critical challenge in symmetric cryptography, often addressed using asymmetric encryption or secure key exchange protocols.

![](https://pbs.twimg.com/media/HQAcG8fWkAApONZ?format=jpg&name=large)

```python
from cryptography.fernet import Fernet
# 1. Generate a key
key = Fernet.generate_key()
print(f"Symmetric Key: {key}\n")
# 2. Initialize the Fernet symmetric encryption object
alice = Fernet(key)
# 3. Encrypt a message
card = b"4676 6390 6332 4255" # b, meaning in bytes
cipher_text = alice.encrypt(card)
print(f"Encrypted Msg. (Ciphertext): {cipher_text}\n")
```

```python
# Decrypt the message
bob = Fernet(key)
decrypted_message = bob.decrypt(cipher_text)
print(f"Decrypted Msg. (Plaintext): {decrypted_message.decode('utf-8')}") # decode from bytes to text
```

## Hashing

**Hashing** is a process that transforms any given input (like a message, file, or password) into a fixed-size string of characters, which is called a hash value, hash code, digest, or simply a hash. This transformation is performed by a hash function, which is a one-way function—meaning it's computationally infeasible to reverse the process and get the original input from the hash value. Good hash functions are designed to be deterministic (the same input always produces the same output) and to minimize collisions (different inputs producing the same output). Hashing is widely used in cybersecurity for data integrity checks (to detect if data has been tampered with), password storage (storing hashes instead of actual passwords), and digital signatures.

```python
import hashlib

# Data to hash
data = '''
Pactum de Mutatione et Solutione

Hoc pactum inter utrasque partes legitime constitutas initum est, quo conventum est ut Promissor, sub condicionibus hic descriptis, summam pecuniae inferius notatam alteri parti persolvat.
Omnia negotia, labores, et officia in hoc contractu definita summa fide et diligentia perficienda sunt, secundum leges iustitiamque vigentes.
Utraque pars consentit omnes controversias futuras per arbitrium boni viri et iustis mediis placide dirimere.

Ad finem huius negotiationis perfectae, debitor vel promissor hic expresse obligatur et spondet se soluturum esse plenam summam
500,000 USD (quingentorum milium dollariorum Civitatum Foederatarum Americae) ad rationem creditoris ante diem constitutum.
Qua solutione integra et rite facta, omnis obligatio legalis inter has partes omnino soluta
et liberata habebitur, nullo alio reclamo manente.
'''

# Update the hash object with the data (encoded as bytes)
hash1= hashlib.sha256(data.encode('utf-8'))

#print(f"Original Data: {data}")
print(f"SHA-256 Hash: {hash1.hexdigest()}") # Get the hexadecimal representation of the hash
```

## Asymmetric Cryptography

**Asymmetric cryptography**, also known as public-key cryptography, uses a pair of mathematically linked keys: a public key and a private key. The public key can be freely distributed and is used for encryption and verifying digital signatures, while the private key is kept secret by its owner and is used for decryption and creating digital signatures. This dual-key system allows for secure communication without the need for a shared secret key, solving the key exchange problem inherent in symmetric cryptography. While more computationally intensive than symmetric encryption, asymmetric cryptography is crucial for secure key exchange, digital signatures, and establishing secure communication channels like SSL/TLS.

![](https://pbs.twimg.com/media/HQAk1QoXgAEGq3E?format=jpg&name=medium)

#### Key pairs generation

```python
from cryptography.hazmat.primitives import serialization
from cryptography.hazmat.primitives.asymmetric import rsa

# Alie key pair!
alicePvtK = rsa.generate_private_key(public_exponent=65537,key_size=1024)
alicePubK = alicePvtK.public_key()
# Export Alice's public key to PEM format and then to hexadecimal
alice_pub = alicePubK.public_bytes(encoding=serialization.Encoding.PEM,format=serialization.PublicFormat.SubjectPublicKeyInfo).hex()
print(f"Alice's Public Key (PEM Hex):\n{alice_pub}\n")

# Bob key pair
bobPvtK = rsa.generate_private_key(public_exponent=65537,key_size=1024)
bobPubK = bobPvtK.public_key()
# Export Bob's public key to PEM format and then to hexadecimal
bob_pub = bobPubK.public_bytes(encoding=serialization.Encoding.PEM,format=serialization.PublicFormat.SubjectPublicKeyInfo).hex()
print(f"Bob's Public Key (PEM Hex):\n{bob_pub}")
```

### Alice:

```python
from cryptography.hazmat.primitives.asymmetric import rsa
from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.primitives.asymmetric import padding

data = b'''
Ad finem ...
500,000 USD ad rationem ....
Que ...
'''

# Sign the message using the private key
sign = alicePvtK.sign(data,padding.PKCS1v15(),hashes.SHA256())
print("Message signed successfully, HASH:", sign)
# Encrypt the content using Bob's public key
toBob = bobPubK.encrypt(data,padding.OAEP(mgf=padding.MGF1(algorithm=hashes.SHA256()),algorithm=hashes.SHA256(),label=None))
print(f"\nPackage encrypted with Bob's public key: {toBob}")
```

#### Bob:

```python
# Bob receives encrypted_package_for_bob
msg = bobPvtK.decrypt(toBob,padding.OAEP(mgf=padding.MGF1(algorithm=hashes.SHA256()),algorithm=hashes.SHA256(),label=None))
print(msg.decode('utf-8'))
# Verify the signature using the public key
try:
    alicePubK.verify(sign,data,padding.PKCS1v15(),hashes.SHA256())
    print("Signature verified successfully! The message is authentic and unaltered.")
except Exception as e:
    print(f"Signature verification failed! Error: {e}")
```
